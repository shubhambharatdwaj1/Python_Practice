```python
import pandas as pd
# print(os.path.exists("D:/DataSets/01_AirlineData/Airline.csv"))

file_path = "c:/Users/Niket/Downloads/Final_Small_Airline.csv"
df = pd.read_csv(file_path)

df
```

```python
for indexnumber , i in df.iterrows():
    print(indexnumber)
    print('-------')
    print(i)
```

```python
for i in df.itertuples():
    print(i)
    for j in i:
        print(j)
    print('--------')
```

```python
for key,Values  in df.items():
    print(key)
    print('-------')
    print(Values)
```

```python
'''Iterrows'''
for index, row in df.iterrows():
    print(index)
    print(row)
    print()
    
```

```python
df
```

```python
'''Using items'''
for colNmae, colValue in df.items():
    print(colNmae)
    print(colValue)
    print()
    print()
```

```python
'''Using itertuples'''
for i in df.itertuples():
    print(i,"\n")
```

```python
'''Using Index'''
for i in df.index:
    print(df['First_name'][i],df['Last_name'][i])
```

```python
for i in df.index:
    print(df.loc[i,['First_name']])
```
