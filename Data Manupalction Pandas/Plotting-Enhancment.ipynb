```python
sns.set_theme(style ="dark", palette = 'pastel')
plt.figure(figsize=(15,5))

Airline_Sort = df.groupby('Airline')['Price'].sum().sort_values().index
ax = sns.barplot(x=df['Airline'] ,y=df['Price'],errorbar=None,estimator = 'sum', legend = False, order = Airline_Sort, alpha =.9,gap =.1)

ax.bar_label(ax.containers[0],rotation = 90)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

sns.set_theme(style="whitegrid", palette='viridis')
plt.figure(figsize=(10,5))

AirLine_sort = df.groupby('Airline')['Price'].count().sort_values().index

ax=sns.countplot(x=df['Airline'], order=AirLine_sort, alpha=0.8, gap=0.3,legend=True)

ax.bar_label(ax.containers[0])
print(ax.containers[0])
plt.xticks(rotation=90)
plt.xlabel("Airline",fontsize=20)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid", palette="RdBu_r")
plt.figure(figsize=(10,5))

AirLine_sort = df.groupby('Airline')['Price'].sum().sort_values().index

ax=sns.countplot(x=df['Airline'], order=AirLine_sort, alpha=0.8, gap=0.3, hue=df['Source'], legend=True, palette='viridis')
ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid", palette='Set1')
plt.figure(figsize=(10,5))

sns.lineplot(x=df['Airline'],y=df['Price'],errorbar=None,marker="s",estimator='mean',alpha=0.9,linestyle="-.")

plt.xticks(rotation=90)
plt.xlabel("Airline",fontsize=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,10))
sns.scatterplot(data=df,x='Price',y = 'Proffit',size='Airline',sizes=(40, 220),hue='Airline',palette="Set2",)
plt.tight_layout()
plt.show()
```

```python
newDf = df.groupby('Airline')['Price'].min().sort_values()
newDf = pd.DataFrame(newDf.reset_index())

sns.barplot(x=newDf['Airline'],y=newDf['Price'])
```

```python
newDF = df.groupby('Airline')['Price'].min().sort_values()
# Convert Series into DataFrame
newDF = newDF.reset_index()

# Create barplot
plt.figure(figsize = (12,6))
sns.barplot(x = newDF['Airline'], y = newDF['Price'], palette = 'viridis')

# Labels and title
plt.xlabel('Airline')
plt.ylabel('minimum Price')
plt.title('Minimum Ticket Price by Airline')

# Rotation x -axis labels
plt.xticks(rotation = 45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (10, 5))
sns.violinplot(y = df['Price'], x = df['Airline'])
plt.show()
```

```python
AirList = df['Airline'].unique()
AirList

plt.figure(figsize = (15, 15))
for key, Value in enumerate(AirList):
    NewDF = df[df['Airline'] == Value]
    plt.subplot(5, 2, key+1)
    sns.violinplot(x = NewDF['Price'])
    plt.tight_layout()
    plt.ylabel(Value)
plt.show()
```

```python
AirList = df['Airline'].unique()
plt.figure(figsize=(20, 20))
for key, value in enumerate(AirList):
    plt.subplot(4, 4, key + 1)
    newDF = df[df['Airline'] == value]
    ax = sns.barplot(data=newDF,x='Airline',y='Price',hue='Source',errorbar=None,estimator='sum')
    for container in ax.containers:
        ax.bar_label(container, fmt='%.0f', padding=3)
        plt.title(value)
    plt.tight_layout()
plt.show()
```
