```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_excel("C:/Users/Niket/Downloads/Swarn_Airline.xlsx")
df.head()
```

```python
df.head(2)
```

```python
order = df.groupby('Airline')['Price'].sum().sort_values().index
order
```

```python
df['Date_of_Journey'] = pd.to_datetime(df['Date_of_Journey'])
df['Date_of_Journey'].dt.month_name()
```

```python
sns.barplot(x=df['Airline'] ,y=df['Price'])
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=df['Airline'] ,y=df['Price'])
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=df['Airline'] ,y=df['Price'])
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=df['Airline'] ,y=df['Price'])
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style ="dark", palette = 'pastel')
plt.figure(figsize=(15,5))

Airline_Sort = df.groupby('Airline')['Price'].sum().sort_values().index
ax = sns.barplot(x=df['Airline'] ,y=df['Price'],errorbar=None,estimator = 'sum', legend = False, order = Airline_Sort, alpha =.9,gap =.1)

ax.bar_label(ax.containers[0],rotation = 90)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

sns.set_theme(style="whitegrid", palette='viridis')
plt.figure(figsize=(10,5))

AirLine_sort = df.groupby('Airline')['Price'].count().sort_values().index

ax=sns.countplot(x=df['Airline'], order=AirLine_sort, alpha=0.8, gap=0.3,legend=True)

ax.bar_label(ax.containers[0])
print(ax.containers[0])
plt.xticks(rotation=90)
plt.xlabel("Airline",fontsize=20)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid", palette="RdBu_r")
plt.figure(figsize=(10,5))

AirLine_sort = df.groupby('Airline')['Price'].sum().sort_values().index

ax=sns.countplot(x=df['Airline'], order=AirLine_sort, alpha=0.8, gap=0.3, hue=df['Source'], legend=True, palette='viridis')
ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid", palette='Set1')
plt.figure(figsize=(10,5))

sns.lineplot(x=df['Airline'],y=df['Price'],errorbar=None,marker="s",estimator='mean',alpha=0.9,linestyle="-.")

plt.xticks(rotation=90)
plt.xlabel("Airline",fontsize=20)
plt.tight_layout()
plt.show()
```
