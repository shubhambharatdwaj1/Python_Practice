```python
class Student:
    def __init__(self):
        self.name = "Rahul"    # Public
        self._age = 20         # Protected
        self.__marks = 90      # Private

    def display(self):
        print(self.name)       # Public
        print(self._age)       # Protected
        print(self.__marks)    # Private

obj = Student()
obj.display()
```

```python
class Student:
    def __init__(self):
        self.name = "Rahul"
        self._age =20
        self.__marks = 90

class Student:
    def display(self):
        print(self.name)
        print(self._age)
        print(self.__marks)

class Teacher:
    def display(self):
        student_Obj = Student()
        print(student_Obj.name)
        print(student_Obj._age)
        print(student_Obj.__marks)

tech_Obj = Teacher()
tech_Obj.display()
```

```python
class Student:
    def display_public(self):
        print("Public Method")

    def _display_procetor(self):
        print("Procetor Method")

    def __display_privet(self):
        print("Privet Method")

    def show(self):
        self.display_public()
        self._display_protected()
        self.__display_private()


obj = Student()
obj.show()
    
```
