```python
def factrial(n):
    if n > 0:
        n = n * factrial(n - 1)
        return n
    else:
        return 1
factrial(10)
```

```python
def cont_up_to(n):
    for i in range(1,n):
        yield i

gen = cont_up_to(n = 10)

print(next(gen))
print(next(gen))
print(next(gen))
```
