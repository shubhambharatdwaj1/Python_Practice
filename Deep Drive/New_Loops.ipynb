```python
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black", "White", "LightGreen"]

for i in col_01:
    for j in col_02:
        if i == j:
            print("Matched", i, j)
```

```python
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black", "White", "LightGreen"]

def func_01(col_01, col_02):
    for i in col_01:
        for j in col_02:
            if i == j:
                return"Matched", i, j
            
vlaues = func_01(col_01, col_02)
vlaues
type(vlaues)
```
