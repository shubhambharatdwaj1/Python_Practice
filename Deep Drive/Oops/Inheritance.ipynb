```python
# Single Inheritance

# Parent Class
class BankAccount:
    def __init__(self, account_holder: str, balance: float = 0.0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount: float):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ₹{amount}. New balance: ₹{self.balance}")
        else:
            print("Deposit amount must be positive.")

    def get_balance(self):
        return f"Account Holder: {self.account_holder} | Balance: ₹{self.balance}"


# Child Class inheriting from BankAccount
class SavingsAccount(BankAccount):
    def __init__(self, account_holder: str, balance: float, interest_rate: float):
        # super().__init__ calls the parent class constructor to set up common attributes
        super().__init__(account_holder, balance)
        # Specific attribute for the child class
        self.interest_rate = interest_rate 

    # Specific method for the child class
    def add_interest(self):
        interest_earned = self.balance * (self.interest_rate / 100)
        self.balance += interest_earned
        print(f"Interest added: ₹{interest_earned:.2f}. New balance: ₹{self.balance:.2f}")


# --- Demonstration ---
# Creating an instance of the child class
savings = SavingsAccount(account_holder="Rahul Sharma", balance=50000.0, interest_rate=4.5)

# Utilizing inherited methods
print(savings.get_balance()) 
savings.deposit(10000.0)

# Utilizing child-specific methods
savings.add_interest()
```

```python
#Single Inheritance

# Parent Class
class InsurancePolicy:
    def __init__(self, policy_number: str, base_premium: float):
        self.policy_number = policy_number
        self.base_premium = base_premium

    def process_billing(self):
        return f"Policy {self.policy_number} processed. Base Premium Due: ₹{self.base_premium}"


# Child Class inheriting from InsurancePolicy
class LifeInsurancePolicy(InsurancePolicy):
    def __init__(self, policy_number: str, base_premium: float, beneficiary: str):
        # Initializing inherited parent attributes
        super().__init__(policy_number, base_premium)
        # Initializing child-specific attribute
        self.beneficiary = beneficiary

    # Specific method for the child class
    def get_policy_summary(self):
        return (f"Life Insurance Policy: {self.policy_number}\n"
                f"Premium Amount: ₹{self.base_premium}\n"
                f"Designated Beneficiary: {self.beneficiary}")


# --- Demonstration ---
# Creating an instance of the child class
life_policy = LifeInsurancePolicy(policy_number="LI-99201", base_premium=12000.0, beneficiary="Anjali Verma")

# Utilizing inherited method
print(life_policy.process_billing())

# Utilizing child-specific method
print(life_policy.get_policy_summary())
```

```python
# Multilevel Inheritance

# Grandparent Class
class BankAccount:
    def __init__(self, account_holder: str, balance: float = 0.0):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount: float):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ₹{amount}. Current Balance: ₹{self.balance}")
        else:
            print("Amount must be positive.")


# Parent Class (Inherits from BankAccount)
class SavingsAccount(BankAccount):
    def __init__(self, account_holder: str, balance: float, interest_rate: float):
        # Initialize grandparent properties
        super().__init__(account_holder, balance)
        self.interest_rate = interest_rate

    def apply_interest(self):
        interest = self.balance * (self.interest_rate / 100)
        self.balance += interest
        print(f"Interest of ₹{interest:.2f} applied. New Balance: ₹{self.balance:.2f}")


# Child Class (Inherits from SavingsAccount)
class PremiumSavingsAccount(SavingsAccount):
    def __init__(self, account_holder: str, balance: float, interest_rate: float, cashback_rate: float):
        # Initialize parent and grandparent properties
        super().__init__(account_holder, balance, interest_rate)
        self.cashback_rate = cashback_rate

    def execute_swipe(self, amount: float):
        """Simulates a debit card swipe that deducts money and awards cashback."""
        if amount <= self.balance:
            self.balance -= amount
            cashback = amount * (self.cashback_rate / 100)
            self.balance += cashback
            print(f"Spent ₹{amount}. Earned ₹{cashback:.2f} cashback! Balance: ₹{self.balance:.2f}")
        else:
            print("Insufficient funds.")


# --- Demonstration ---
# Creating an instance of the bottom-tier child class
premium_user = PremiumSavingsAccount(
    account_holder="Karan Johar", 
    balance=100000.0, 
    interest_rate=5.0, 
    cashback_rate=1.5
)

# 1. Using Grandparent method
premium_user.deposit(50000.0) 

# 2. Using Parent method
premium_user.apply_interest() 

# 3. Using Child method
premium_user.execute_swipe(20000.0) 
```

```python
# Multilevel inheritance

# Grandparent Class
class InsurancePolicy:
    def __init__(self, policy_id: str, base_premium: float):
        self.policy_id = policy_id
        self.base_premium = base_premium

    def get_billing_info(self):
        return f"Policy: {self.policy_id} | Annual Premium Due: ₹{self.base_premium}"


# Parent Class (Inherits from InsurancePolicy)
class LifeInsurance(InsurancePolicy):
    def __init__(self, policy_id: str, base_premium: float, beneficiary: str):
        super().__init__(policy_id, base_premium)
        self.beneficiary = beneficiary

    def get_beneficiary_details(self):
        return f"Designated Beneficiary for {self.policy_id} is: {self.beneficiary}"


# Child Class (Inherits from LifeInsurance)
class TermLifeInsurance(LifeInsurance):
    def __init__(self, policy_id: str, base_premium: float, beneficiary: str, duration_years: int):
        super().__init__(policy_id, base_premium, beneficiary)
        self.duration_years = duration_years

    def get_full_summary(self):
        # Accessing fields from Grandparent, Parent, and Child simultaneously
        return (f"--- Policy Contract Overview ---\n"
                f"ID: {self.policy_id} (Active for {self.duration_years} Years)\n"
                f"Cost: ₹{self.base_premium}/year\n"
                f"Payout Target: {self.beneficiary}")


# --- Demonstration ---
# Creating an instance of the bottom-tier child class
term_policy = TermLifeInsurance(
    policy_id="TLI-2026-883", 
    base_premium=15000.0, 
    beneficiary="Priya Patel", 
    duration_years=25
)

# Accessing methods up the multilevel inheritance ladder
print(term_policy.get_billing_info())          # Grandparent layer
print(term_policy.get_beneficiary_details())    # Parent layer
print(term_policy.get_full_summary())           # Child layer
```
