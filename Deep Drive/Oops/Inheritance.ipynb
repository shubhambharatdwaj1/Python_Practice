```python
class Certificate:
    def __init__(self, cert_name):
        self.cert_name = cert_name

    def print_certificate(self):
        print(f"Certificate: {self.cert_name}")

# Composition
class Course:
    def __init__(self, course_name, cert_name):
        self.course_name = course_name
        self.certificate = Certificate(cert_name)  

    def course_details(self):
        print(f"Course: {self.course_name}")
        self.certificate.print_certificate()

course = Course("Data Science", "Intellipaat Certified Data Scientist")
course.course_details()
```

```python
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def display_info(self):
        print(f"Name: {self.name}, ID: {self.emp_id}")

class TechnicalWriter(Employee):
    def __init__(self, name, emp_id, specialty):
        super().__init__(name, emp_id)
        self.specialty = specialty

    # Overriding the method
    def display_info(self):
        print(f"Name: {self.name}, ID: {self.emp_id}, Specialty: {self.specialty}")
        
writer = TechnicalWriter("Shubham", 303, "Data Science")
writer.display_info()
```

```python
class Intellipaat:
    def __init__(self, org_name):
        self.org_name = org_name

    def org_info(self):
        return f"Organization: {self.org_name}"

# Single Inheritance
class Employee(Intellipaat):
    def __init__(self, org_name, name, emp_id):
        super().__init__(org_name)
        self.name = name
        self.emp_id = emp_id

    def display(self):
        return f"Name: {self.name}, ID: {self.emp_id}"

class TechnicalWriter(Employee):
    def write_docs(self):
        return f"{self.name} writes documentation."

class Sales(Employee):
    def handle_clients(self):
        return f"{self.name} handles clients."

# Multilevel Inheritance
class Course(Intellipaat):
    def __init__(self, org_name, title, duration):
        super().__init__(org_name)
        self.title = title
        self.duration = duration

    def course_info(self):
        return f"Course: {self.title}, Duration: {self.duration}"

class ProgrammingCourse(Course):
    def language(self):
        return "Covers Python, JavaScript, etc."

class FullStackCourse(ProgrammingCourse):
    def stack_info(self):
        return "Includes frontend, backend, and deployment."

tw = TechnicalWriter("Intellipaat", "Shubham", 101)
print(tw.display())
print(tw.write_docs())
print(tw.org_info())

fs = FullStackCourse("Intellipaat", "Full Stack Dev", "6 Months")
print(fs.course_info())
print(fs.language())
print(fs.stack_info())
print(fs.org_info())

```

```python
# Parent class
class Bank:
    def bank_details(self):
        print("Bank Name: State Bank of India")

# Child class
class Banking(Bank):
    def savings_account(self):
        print("Account Type: Savings Account")

    def deposit(self, amount):
        print("Deposited Amount:", amount)


# Create object
b = Banking()

# Call parent class method
b.bank_details()

# Call child class methods
b.savings_account()
b.deposit(5000)
```

```python
# Multilevel Inheritance

# Grandparent class
class Bank:
    def bank_name(self):
        print("Bank Name: SBI")

# Parent class
class Account(Bank):
    def account_type(self):
        print("Account Type: Savings Account")

# Child class
class Customer(Account):
    def customer_name(self):
        print("Customer Name: Rahul")


# Create object of child class
c = Customer()

# Access methods from all three levels
c.bank_name()
c.account_type()
c.customer_name()
```

```python
# Multilevel Inheritance

# Grandparent class
class Insurance:
    def company_name(self):
        print("Insurance Company: LIC")


# Parent class
class LifeInsurance(Insurance):
    def policy_type(self):
        print("Policy Type: Life Insurance")


# Child class
class Customer(LifeInsurance):
    def customer_details(self):
        print("Customer Name: Rahul")
        print("Annual Premium: Rs. 12000")


# Create object
c = Customer()

# Call methods from all three classes
c.company_name()
c.policy_type()
c.customer_details()
```
