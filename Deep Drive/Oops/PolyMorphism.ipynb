# Method Overloding

```python
class myClass:
    def sum(self, a = None, b =None, c= None, d =None):
        if a != None and b !=None and c != None and d!=None :
           return a+b+c+d
        if a!= None and b != None and c!= None:
            return a+b+c
        elif a!=None and b!= None:
            return a+b
        else:
            return "Provide atleast 2 argument"

Obj = myClass()
sum = Obj.sum(5,10,1,6)
print(sum)
```

# Method Overloding

```python
class Animal:
    def sound(self):
        return "Animal makes sound !"

class Dog(Animal):
    def sound(self):
        return "Dog Barks !"

class Cat(Animal):
    def sound(self):
        return "Cat Meaw !"

AniObj= Animal()
DogObj = Dog()
CatObj = Cat()

print(AniObj.sound())
print(DogObj.sound())
print(CatObj.sound())
```

```python
class P_C_Intrested:
    def instrested(self):
        return 0.011

class C_C_Home__Instrested(P_C_Intrested):
    def instrested(self):
        return 0.08

Home_Obj = C_C_Home__Instrested()
Home_Obj.instrested()


```
