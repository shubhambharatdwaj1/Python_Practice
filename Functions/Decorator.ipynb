```python
def my_devi_decor(func):
    def wrapper(x,y):
        if isinstance(x, str):
            x = int(x)
        if isinstance(y, str):
            y = int(y)
            func(x,y)
        if y>x:
            x,y = y,x
            print("Value Swapped")
        return func(x,y)
    return wrapper

@my_devi_decor
def OriginalFun(x,y):
    print(x/y)
OriginalFun('5',10)
```

```python
List_01 = ['1', '2', '3', '4']
List_02 = ['10', '20', '30', '40']

# 1. Define the decorator to process strings and return a dictionary
def process_and_store_dict(func):
    def wrapper(lst):
        # Converts string values to integers and calculates the square
        result_dict = {item: int(item) ** 2 for item in lst}
        return result_dict
    return wrapper

# 2. Apply the decorator to your original function
@process_and_store_dict
def original_fun(List_01):
    for i in List_01:
        print(i ** 2)

# 3. Execute the function and print the dictionary output
output_01 = original_fun(List_01)
output_02 = original_fun(List_02)

print("Dictionary 1:", output_01)
print("Dictionary 2:", output_02)
```
