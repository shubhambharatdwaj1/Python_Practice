```python
def my_devi_decor(func):
    def wrapper(x,y):
        if isinstance(x, str):
            x = int(x)
        if isinstance(y, str):
            y = int(y)
            func(x,y)
        if y>x:
            x,y = y,x
            print("Value Swapped")
        return func(x,y)
    return wrapper

@my_devi_decor
def OriginalFun(x,y):
    print(x/y)
OriginalFun('5',10)
```

```python
List_01 = ['1', '2', '3', '4']
List_02 = ['10', '20', '30', '40']

def store_in_dict(func):
    def wrapper(values):
        return {int(value): func(int(value)) for value in values}
    return wrapper

@store_in_dict
def original_fun(value):
    return value ** 2

result = original_fun(List_01)
print(result)  # {1: 1, 2: 4, 3: 9, 4: 16}
print(original_fun(List_02))
# {10: 100, 20: 400, 30: 900, 40: 1600}

```

```python
List_01 = ['1', '2', '3', '4']
List_02 = ['10', '20', '30', '40']

# 1. Define the decorator to process strings and return a dictionary
def process_and_store_dict(func):
    def wrapper(lst):
        # Converts string values to integers and calculates the square
        result_dict = {item: int(item) ** 2 for item in lst}
        return result_dict
    return wrapper

# 2. Apply the decorator to your original function
@process_and_store_dict
def original_fun(List_01):
    for i in List_01:
        print(i ** 2)

# 3. Execute the function and print the dictionary output
output_01 = original_fun(List_01)
output_02 = original_fun(List_02)

print("Dictionary 1:", output_01)
print("Dictionary 2:", output_02)
```

```python
def decorator(func):
    def wrapper():
        print("Before calling the function.")
        func()
        print("After calling the function.")
    return wrapper

@decorator
def greet():
    print("Hello, World!")
greet()
```

```python
def decorator_name(func):
    def wrapper(*args, **kwargs):
        print("Before execution")
        result = func(*args, **kwargs)
        print("After execution")
        return result
    return wrapper

@decorator_name
def add(a, b):
    return a + b

print(add(5, 3))
```

```python
# Assigning a function to a variable
def greet(n):
    return f"Hello, {n}!"
say_hi = greet 
print(say_hi("Alex")) 

# Passing a function as an argument
def apply(f, v):
    return f(v)
res = apply(say_hi, "Elon")
print(res) 

# Returning a function from another function
def make_mult(f):
    def mult(x):
        return x * f
    return mult
dbl = make_mult(2)
print(dbl(5))
```

```python
def method_decorator(func):
    def wrapper(self, *args, **kwargs):
        print("Before method execution")
        res = func(self, *args, **kwargs)
        print("After method execution")
        return res
    return wrapper

class MyClass:
    @method_decorator
    def say_hello(self):
        print("Hello!")
obj = MyClass()
obj.say_hello()
```

```python
def fun(cls):
    cls.class_name = cls.__name__
    return cls

@fun
class Person:
    pass
print(Person.class_name)
```

```python
class MathOperations:
    @staticmethod
    def add(x, y):
        return x + y

res = MathOperations.add(5, 3)
print(res)
```

```python
class Employee:
    raise_amount = 1.05
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        
    @classmethod
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount

Employee.set_raise_amount(1.10)
print(Employee.raise_amount)
```
