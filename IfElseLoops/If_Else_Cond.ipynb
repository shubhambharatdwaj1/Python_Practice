```python
# IF condition
# check Value
# please check if the value is less than or equal to 10

i = 12
if i < 5:
    print("Given value is less than 5")
elif i < 8:
    print("Given values is less than 8")
elif i < 12:
    print("Given Values is less than 12")
elif i == 12:
    print("Given Value equal to 12")
else:
    print("Given Value is Greater than 12")
```
