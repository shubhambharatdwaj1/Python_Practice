```python
# Intitialize the variable 
counter = 5

# The loop runs as long as counter
while counter >= 0:
    print (f"counter is : {counter}")
    counter -= 1    #Decrease counter
```

```python
'''Nested Loop'''
# Find the matching Element from two lists
List_1 = ["Red", "Yellow" "LightGreen"]
List_2 = ["white", "Black", "Yellow"]

for i in List_1 :
    for j in List_2 :
        if i==j :
            print("Matched", i,j,)
```
