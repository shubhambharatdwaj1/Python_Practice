```python
outer_counter = 3

while outer_counter > 0 :
    print(f"Outer loop: {outer_counter}")

    inner_counter = 2

    while inner_counter > 0:
        print(f"Inner loop: {inner_counter}")
        inner_counter -= 1

    outer_counter -= 1
    print()
```
