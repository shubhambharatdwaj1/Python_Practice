```python
# Intitialize the variable 
counter = 5

# The loop runs as long as counter
while counter >= 0:
    print (f"counter is : {counter}")
    counter -= 1    #Decrease counter
```

```python
# Intitialize the variable
number = 1

# The loop runs as long as number i
while number != 4:
    print(f"Number is: {number}")
    number += 1    # Increase number
```
