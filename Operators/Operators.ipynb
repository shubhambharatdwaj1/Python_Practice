```python
result = (150 <= 1000) or (20 >= 150)
result
```

```python
result = (20 > 15) and (25 > 20)

print(result)
```

```python
'''Logical AND (and): Returns True if both operands are True.'''
# True and True = True
# True and False = False
# False and True = False
# False and False = False

result = (15 > 15) and (15 > 20)

print(result)
```

```python
result = (150 > 400) or (117 > 2000)

print(result)
```

```python
'''Logical OR (or): Returns True if at least one of the operands is True.'''
# True or True = True
# True or False = True
# False or True = True
# False 0r False = False

result =(5 > 4) or (117 > 20)
print(result)
```

```python
'''Logical NOT (not): Returns the opposite of the operand's truth value.'''

result = not (10 < 0)
print(result)
```
