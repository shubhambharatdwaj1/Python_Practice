```python
file_path = "c:/Users/Niket/Downloads/Airline1 (3).csv"
df = pd.read_csv(file_path)
df
```

```python
Source_value = df['Source'].mode()[0]
df['Source'] = df['Source'].fillna(Source_value)
df['Source']
```

```python
Destination_Value = df['Destination'].mode()[0]
df['Destination'] = df['Destination'].fillna(Destination_Value)
df['Destination']
```

```python
Route_Value = df['Route'].mode()[0]
df['Route'] = df['Route'].fillna(Route_Value)
df['Route']
```

```python
Total_Stops_Value = df['Total_Stops'].mode()[0]
df['Total_Stops'] = df['Total_Stops'].fillna(Total_Stops_Value)
df['Total_Stops']
```

```python
Price_Value = df['Price'].mean()
df['Price'] = df['Price'].fillna(Price_Value)
df['Price']
```

```python
Destination_Value = df['Destination'].mode()[0]
df['Destination'] = df['Destination'].fillna(Destination_Value)
df['Destination']
Route_Value = df['Route'].mode()[0]
df['Route'] = df['Route'].fillna(Route_Value)
df['Route']
Total_Stops_Value = df['Total_Stops'].mode()[0]
df['Total_Stops'] = df['Total_Stops'].fillna(Total_Stops_Value)
df['Total_Stops']
Price_Value = df['Price'].mean()
df['Price'] = df['Price'].fillna(Price_Value)
df['Price']
```

```python
df['Airline'] = df['Airline'].replace("vistars Premium encomy", "vistara")
df['Destination'] = df['Destination'].replace("New Delhi","Delhi")
df['Destination'].value_counts()
```

# Graph Basesd Aanalysis

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x = df['Airline'])
plt.show()
```

```python
sorted = df.groupby('Airline')['Price'].count().sort_values().index
ax = sns.countplot(x = df['Airline'], order = sorted)
plt.xticks(rotation = 90)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sorted = df.groupby('Airline')['Price'].count().sort_values().index
ax = sns.countplot(x = df['Airline'], order = sorted)
ax.bar_label(ax.containers[0])
plt.xticks(rotation = 90)
plt.show()
```
