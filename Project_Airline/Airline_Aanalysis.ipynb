```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x = df['Airline'])
plt.show()
```

```python
sorted = df.groupby('Airline')['Price'].count().sort_values().index
ax = sns.countplot(x = df['Airline'], order = sorted)
plt.xticks(rotation = 90)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sorted = df.groupby('Airline')['Price'].count().sort_values().index
ax = sns.countplot(x = df['Airline'], order = sorted)
ax.bar_label(ax.containers[0])
plt.xticks(rotation = 90)
plt.show()
```

```python
sorted = df.groupby('Source')['Price'].count().sort_values().index
ax=sns.countplot(x=df['Source'],order=sorted)
ax.bar_label(ax.containers[0])
plt.xticks(rotation=90)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sorted = df.groupby('Total_Stops')['Price'].count().sort_values().index
ax = sns.countplot(x = df['Total_Stops'], order = sorted)
ax.bar_label(ax.containers[0])
plt.xticks(rotation = 90)
plt.show()
```

```python
columns = ['Total_Stops', 'Source', 'Airline']

plt.figure(figsize=(15,5))
for key,value in enumerate(columns):
    order = df[value].value_counts().sort_values().index
    ax=sns.countplot(data=df, x=value, order=order)
    plt.xlabel("Total_Stops, Source, Airline")
    # ax.bar_label(ax.containers[0])
    for i in ax.containers:
        ax.bar_label(i, fmt='%.0f', size=8)
    plt.xticks(rotation = 90)
```

```python
columns = ['Total_Stops', 'Source', 'Airline']

plt.figure(figsize=(15,5))
for key,value in enumerate(columns):
    order = df[value].value_counts().sort_values().index
    ax=sns.barplot(data=df, x=value,y = 'Price', order=order, errorbar = None)
    plt.xlabel("Total_Stops, Source, Airline")
    # ax.bar_label(ax.containers[0])
    for i in ax.containers:
        ax.bar_label(i, fmt='%.0f', size=8)
    plt.xticks(rotation = 90)
plt.show()
```

```python
aggList = ["mean","sum","min","max"]
plt.figure(figsize=(15,10))
for key, value in enumerate(aggList):
    newDf = df.groupby('Airline')['Price'].agg(value).sort_values()
    newDf = pd.DataFrame(newDf.reset_index())
    plt.subplot(2,2,key+1)
    ax = sns.barplot(data =newDf,x = 'Airline',y = "Price")
                
    ax.bar_label(ax.containers[0],rotation =90)
    plt.xticks(rotation =90)
    plt.title(value)
plt.tight_layout()
plt.show()

```

```python
aggList = ["mean","sum","min","max"]
plt.figure(figsize=(15,10))
for key, value in enumerate(aggList):
    newDf = df.groupby('Total_Stops')['Price'].agg(value).sort_values()
    newDf = pd.DataFrame(newDf.reset_index())
    plt.subplot(2,2,key+1)
    ax = sns.barplot(data =newDf,x = 'Total_Stops',y = "Price")
                
    ax.bar_label(ax.containers[0],rotation =90)
    plt.xticks(rotation =90)
    plt.title(value)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,10))
for key,value in enumerate(aggList):
    newdf = df.groupby('Source')['Price'].agg(value).sort_values()
    newdf = pd.DataFrame(newdf.reset_index())
    plt.subplot(2,2,key+1)

    ax = sns.barplot(data=newdf, x='Source', y='Price')
    ax.bar_label(ax.containers[0], size=10)
    plt.xticks(rotation=90)
    plt.title(value)
plt.tight_layout()
plt.show()
```

```python
newDf = df.groupby("Total_Stops")["Price"].agg(['sum','count']).sort_values(by ="sum")
newDf = newDf.reset_index()
ax = sns.barplot(data = newDf,x ='Total_Stops',y = 'sum', palette='Set3')
ax.twinx()
ax2 = sns.lineplot(data =newDf,x = 'Total_Stops',y = 'count', color ="yellow", markers="o")
```

```python
df['Sector'] =df['Source']+"_"+df['Destination']
df

Secto_df =df.groupby(['Total_Stops','Sector'])['Price'].mean()
Secto_df = Secto_df.reset_index()
Secto_df

new_df = Secto_df[Secto_df['Total_Stops'] == '1 stop']
new_df
sns.barplot(x = new_df['Sector'],y = new_df['Price'])
plt.title('1 stop')
plt.show()
```

```python
STOPS_List = ['non-stop',  '1 stop', '2 stops','3 stops', '4 stops']

plt.figure(figsize=(20,12))

for k,v in enumerate(STOPS_List):

    new_df = Secto_df[Secto_df['Total_Stops']==v]
    plt.subplot(3,2,k+1)
    ax = sns.barplot(x=new_df['Sector'],y=new_df['Price'])
    ax.bar_label(ax.containers[0])
    plt.xticks(rotation=10)
    plt.title(v)
    plt.subplots_adjust(hspace=0.9)
plt.show()
```
